```python

#Q1. A list holds the maximum temperatures (in Celsius) of Mumbai for 7 days.
 Write a program to create a second list with the Fahrenheit equivalents
  (F = C * 9/5 + 32), print both side by side in a formatted table, and display
the average, the hottest day number and a tuple (minimum, maximum) of
 the Celsius readings.
# Maximum temperatures of Mumbai for 7 days
celsius = [30, 32, 31, 33, 35, 34, 29]

# Create a second list for Fahrenheit temperatures
fahrenheit = []

for temp in celsius:
    f = temp * 9/5 + 32
    fahrenheit.append(f)

# Display both temperatures in a formatted table
print("Day\tCelsius (°C)\tFahrenheit (°F)")
print("-" * 40)

for i in range(7):
    print(f"{i+1}\t{celsius[i]}\t\t{fahrenheit[i]:.2f}")

# Calculate average Celsius temperature
average = sum(celsius) / len(celsius)
print("\nAverage Celsius temperature:", round(average, 2), "°C")

# Find the hottest day
hottest = celsius.index(max(celsius)) + 1
print("Hottest day number:", hottest)

# Find minimum and maximum Celsius temperatures
temperature_range = (min(celsius), max(celsius))
print("Minimum and maximum:", temperature_range)
```

```python

#Q2. Write a menu-driven calculator program with separate functions add(),
 subtract(), multiply() and divide(). Use a while loop to keep showing the menu
 until the user chooses Exit, accept two numbers on each turn, and display the
 result formatted to 2 decimals. Division by zero must be handled with an if
 statement and a suitable message.
# Menu-driven calculator using functions

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        print("Error: Division by zero is not allowed!")
        return None
    else:
        return a / b

# Menu repeats until the user chooses Exit
while True:
    print("\n----- CALCULATOR MENU -----")
    print("1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Divide")
    print("5. Exit")

    choice = input("Enter your choice (1-5): ")

    if choice == "5":
        print("Calculator closed. Goodbye!")
        break

    if choice in ("1", "2", "3", "4"):
        num1 = float(input("Enter first number: "))
        num2 = float(input("Enter second number: "))

        if choice == "1":
            result = add(num1, num2)
        elif choice == "2":
            result = subtract(num1, num2)
        elif choice == "3":
            result = multiply(num1, num2)
        elif choice == "4":
            result = divide(num1, num2)

        if result is not None:
            print(f"Result: {result:.2f}")
    else:
        print("Invalid choice! Please enter a number from 1 to 5.")
```
